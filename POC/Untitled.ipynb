```python
import mne
import numpy as np
import matplotlib.pyplot as plt
import base64
import gzip
import numpy as np
import io
def load_edf(file_path):
    """
    Load an EDF file and return the raw EEG object.
    """
    raw = mne.io.read_raw_edf(file_path, preload=True)
    return raw

def select_channels(raw, channel_list):
    """
    Select a subset of EEG channels for processing.
    """
    raw_selected = raw.copy().pick(channel_list)
    return raw_selected

def set_standard_montage(raw):
    """
    Assigns the standard 10-20 electrode layout to the raw EEG data.
    """
    montage = mne.channels.make_standard_montage('standard_1020')
    raw.set_montage(montage)
    return raw

def set_custom_montage(raw, montage_path):
    """
    Loads and assigns a custom montage from a given file path.
    """
    montage = mne.channels.read_custom_montage(montage_path)
    raw.set_montage(montage)
    return raw

def interpolate_bad_channels(raw):
    """
    Interpolates bad EEG channels if they exist.
    """
    if raw.info['bads']:  # Only interpolate if bad channels exist
        raw.interpolate_bads(reset_bads=True)
    else:
        print("No bad channels found. Skipping interpolation.")
    return raw

def plot_raw(raw):
    """
    Plot raw EEG data interactively.
    """
    plt.ion()
    raw.plot(block=True)

def save_processed_data(raw, output_dir, trial_name, file_extension=".fif"):
    """
    Save the processed EEG data to a specified directory.
    """
    file_name = f"{output_dir}/{trial_name}{file_extension}"
    raw.save(file_name, overwrite=True)
    return file_name

# Example Usage
if __name__ == "__main__":
    file_path = "P2_T2.edf"  # Change this to the actual EDF file path
    output_dir = "output"  # Change this to your desired output folder
    trial_name = "processed_eeg"
    channels_8 = ['AF3', 'F7', 'F3', 'FC5', 'FC6', 'F4', 'F8', 'AF4']
    
    raw = load_edf(file_path)
    # raw = select_channels(raw, channels_8)
    # raw = set_standard_montage(raw)  # or set_custom_montage(raw, "montage_file.elp")
    # raw = interpolate_bad_channels(raw)
    
    # Optional plotting for debugging
    # plot_raw(raw)
    
    # save_path = save_processed_data(raw, output_dir, trial_name)
    # print(f"Processed data saved at: {save_path}")
```
