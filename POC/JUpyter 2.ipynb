```python
import mne
import matplotlib.pyplot as plt
import numpy as np
import matplotlib
matplotlib.use('Qt5Agg')
raw = mne.io.read_raw_edf("P2_T1.edf", preload=True)
selected_channels= raw.ch_names[6:20]
# Pick the selected channels

raw_selected = raw.copy().pick_channels(selected_channels)

# Get time values (x-axis)
times = raw_selected.times  

# Get data values (y-axis)
data, _ = raw_selected[:, :]  

# Scale the data (convert EEG from volts to microvolts)
data *= 1e6  # Convert V to µV for better visibility

# Number of selected channels
n_channels = len(selected_channels)

# Offset for stacking signals
offset = np.max(np.abs(data)) * 1.5  # Spacing factor
y_offsets = np.arange(n_channels) * offset  

# Plot the selected channels
plt.figure(figsize=(12, 6))
for i, ch_name in enumerate(selected_channels):
    plt.plot(times, data[i] + y_offsets[i], label=ch_name, linewidth=0.7)

# Formatting
plt.xlabel('Time (seconds)')
plt.ylabel('Amplitude (µV, arbitrary offset)')
plt.title('Time Series Data for Selected Channels')
plt.legend(loc='upper right', fontsize='small')
plt.grid(True)

# Show the plot
plt.show()
```
